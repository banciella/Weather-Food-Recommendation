# 03 - Setup do ambiente

Cria a estrutura base do projeto no Unity Catalog (catálogo `workspace`):

- Schemas `bronze`, `silver` e `gold` (arquitetura Medalhão)
- Volume `bronze.raw_files`, onde ficam os JSONs brutos das APIs

Roda uma vez e é seguro rodar de novo (`IF NOT EXISTS`). A segunda célula grava e lê um arquivo de teste no Volume, para confirmar a permissão de escrita.
```

````python
CATALOG = "workspace"

for schema in ["bronze", "silver", "gold"]:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{schema}")

spark.sql(f"CREATE VOLUME IF NOT EXISTS {CATALOG}.bronze.raw_files")

display(spark.sql(f"SHOW SCHEMAS IN {CATALOG}"))
````

````python
path = "/Volumes/workspace/bronze/raw_files/teste.txt"

dbutils.fs.put(path, "volume ok", overwrite=True)
print(dbutils.fs.head(path))
````
